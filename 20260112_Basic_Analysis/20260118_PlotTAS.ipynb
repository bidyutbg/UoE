```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(
    nrows=len(MODELS),
    ncols=1,
    figsize=(10, 2.5 * len(MODELS)),
    sharex=True,
)

axes = np.atleast_1d(axes)

for ax, (model_name, model_meta) in zip(axes, MODELS.items()):

    var = {}
    var_by_year = {}

    # --- build paths + read data ---
    for exp, meta in EXPERIMENTS.items():

        # --- special-case ensemble override ---
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ensemble = "r1i1p1f2"
            else:
                ensemble = "r1i1p1f1"
        else:
            ensemble = model_meta["ensemble"]


        base = (
            CEDA_BASE
            / meta["project"]
            / model_meta["institution"]
            / model_name
            / meta["scenario"]
            / ensemble
            / "Amon"
            / varname
            / model_meta["grid"]
            / "latest"
        )

        print(str(base))
        # ds = mf.open_files(str(base))
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ds = mf.open_files_CESM_G6sulfur(base)
            elif meta["scenario"] == "ssp585":
                ds = mf.open_files_CESM_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        elif model_name == "IPSL-CM6A-LR":
            if meta["scenario"] == "ssp585":
                ds = mf.open_files_IPSL_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        else:
            ds = mf.open_files(str(base))

        var[exp] = mf.read_var(ds, varname)

    # --- seasonal means (KEEP seasons dict as-is) ---
    seasons = {
        "ANN":  (1, 12),
        # "JJAS": (6, 9),
        # "DJF":  (12, 2),
    }

    for exp, da in var.items():
        var_by_year[exp] = {}
        for season, (m1, m2) in seasons.items():
            var_by_year[exp][season] = mf.seasonal_mean_by_year(da, m1, m2)

    # --- plotting (ANN only for now) ---
    for exp, meta in EXPERIMENTS.items():
        ts = var_by_year[exp]["ANN"].mean(dim=("lat", "lon"))
        ax.plot(
            ts.year,
            ts,
            label=exp,
            color=meta.get("color", None),
        )

    ax.set_ylabel(model_name, fontsize=9)
    ax.set_ylim(275,285.5)
    # ax.grid(True)
    ax.axhline(y=277, color='gray', linestyle='--')
    ax.axvspan(2021, 2050, color='lightpink', alpha=0.3)
    ax.axvspan(2071, 2100, color='lightblue', alpha=0.3)

axes[-1].set_xlabel("Year")
axes[0].legend(ncol=5, fontsize=9, loc="upper left")

plt.suptitle("Annual Global Mean Surface Temperature", y=0.93)
plt.xlim(1850, 2100)

# Save
fig_directory = "/home/users/bidyut/UoE/20260112_Basic_Analysis"
figname = "Figure_GMST_AllModels_PI2PD2SAI"
# # plt.savefig(''+str(fig_directory)+'/'+str(figname)+'.svg', format="svg",transparent=True, dpi=1200)
# # !rsvg-convert -f pdf -o {fig_directory}/{figname}.pdf {fig_directory}/{figname}.svg
plt.savefig(f"{fig_directory}/{figname}.pdf", format="pdf")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

fig, ax = plt.subplots(figsize=(16, 8))

# Dictionary to store all model data by experiment
data_by_experiment = {exp: [] for exp in EXPERIMENTS.keys()}

# --- Collect data from all models ---
for model_name, model_meta in MODELS.items():
    var = {}
    var_by_year = {}
    
    # --- build paths + read data ---
    for exp, meta in EXPERIMENTS.items():
        # --- special-case ensemble override ---
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ensemble = "r1i1p1f2"
            else:
                ensemble = "r1i1p1f1"
        else:
            ensemble = model_meta["ensemble"]
        
        base = (
            CEDA_BASE
            / meta["project"]
            / model_meta["institution"]
            / model_name
            / meta["scenario"]
            / ensemble
            / "Amon"
            / varname
            / model_meta["grid"]
            / "latest"
        )
        print(str(base))
        
        # Read data with model-specific handlers
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ds = mf.open_files_CESM_G6sulfur(base)
            elif meta["scenario"] == "ssp585":
                ds = mf.open_files_CESM_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        elif model_name == "IPSL-CM6A-LR":
            if meta["scenario"] == "ssp585":
                ds = mf.open_files_IPSL_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        else:
            ds = mf.open_files(str(base))
        
        var[exp] = mf.read_var(ds, varname)
    
    # --- seasonal means ---
    seasons = {
        "ANN": (1, 12),
    }
    
    for exp, da in var.items():
        var_by_year[exp] = {}
        for season, (m1, m2) in seasons.items():
            var_by_year[exp][season] = mf.seasonal_mean_by_year(da, m1, m2)
    
    # --- Store global mean time series for each experiment ---
    ensemble_means = {}
    for exp in EXPERIMENTS.keys():
        ts = var_by_year[exp]["ANN"].mean(dim=("lat", "lon"))
        data_by_experiment[exp].append(ts)

# --- Define common year range ---
common_years = np.arange(1850, 2101)  # Adjust as needed

# --- Plot ensemble mean and spread ---
for exp, meta in EXPERIMENTS.items():
    # Stack all model data for this experiment
    all_models = data_by_experiment[exp]
    
    if len(all_models) == 0:
        continue
    
    # Align all models to common year grid
    aligned_data = []
    for ts in all_models:
        # Create a pandas Series for easy reindexing
        series = pd.Series(ts.values, index=ts.year.values)
        # Reindex to common years (NaN where data doesn't exist)
        aligned = series.reindex(common_years)
        aligned_data.append(aligned.values)
    
    # Stack into 2D array (models x years)
    model_data = np.array(aligned_data)
    
    # Calculate ensemble mean and spread (ignoring NaNs)
    ensemble_mean = np.nanmean(model_data, axis=0)
    ensemble_std = np.nanstd(model_data, axis=0)
    
    # Only plot where we have at least one model's data
    valid_mask = ~np.isnan(ensemble_mean)
    
    # Save ensemble_mean
    ensemble_means[exp] = ensemble_mean
    # Plot mean line
    line_color = meta.get("color", None)
    ax.plot(
        common_years[valid_mask],
        ensemble_mean[valid_mask],
        label=exp,
        color=line_color,
        linewidth=2,
    )
    
    # Plot shaded spread (±1 standard deviation)
    ax.fill_between(
        common_years[valid_mask],
        (ensemble_mean - ensemble_std)[valid_mask],
        (ensemble_mean + ensemble_std)[valid_mask],
        facecolor=line_color,  # Use facecolor instead of color
        alpha=0.3,
        edgecolor='none',  # Remove edge color
    )

ax.set_ylabel("Temperature (K)", fontsize=11)
ax.set_xlabel("Year", fontsize=11)
ax.set_ylim(275, 287)
# ax.axvspan(2021, 2050, color='lightpink', alpha=0.3)
# ax.axvspan(2071, 2100, color='lightblue', alpha=0.3)
ax.legend(ncol=5, fontsize=9, loc="upper left")
ax.set_title("Annual Global Mean Surface Temperature - Multi-Model Ensemble", fontsize=12)
plt.xlim(1850, 2100)
ax.grid(True, alpha=0.3)

# Save
fig_directory = "/home/users/bidyut/UoE/20260112_Basic_Analysis"
figname = "Figure_GlobalMeanSfcTemp_EnsembleMean_PI2PD2SAI"
plt.savefig(f"{fig_directory}/{figname}.pdf", format="pdf", bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import warnings

# Create figure with one subplot per experiment
fig, axes = plt.subplots(
    nrows=len(EXPERIMENTS),
    ncols=1,
    figsize=(10, 2.5 * len(EXPERIMENTS)-1),
    sharex=True,
)
axes = np.atleast_1d(axes)

# Storage for all model data by experiment
data_by_experiment = {exp: {'global': [], 'land': [], 'ocean': []} 
                      for exp in EXPERIMENTS.keys()
                      if exp != "HIST"
                     }

# First pass: collect data from all models for each experiment
for model_name, model_meta in MODELS.items():
    print(f"Processing model: {model_name}")
    
    for exp, meta in EXPERIMENTS.items():
        if exp == "HIST":
            continue
        # --- special-case ensemble override ---
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ensemble = "r1i1p1f2"
            else:
                ensemble = "r1i1p1f1"
        else:
            ensemble = model_meta["ensemble"]
        
        base = (
            CEDA_BASE
            / meta["project"]
            / model_meta["institution"]
            / model_name
            / meta["scenario"]
            / ensemble
            / "Amon"
            / varname
            / model_meta["grid"]
            / "latest"
        )
        print(f"  {exp}: {str(base)}")
        
        # Open files with model-specific logic
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ds = mf.open_files_CESM_G6sulfur(base)
            elif meta["scenario"] == "ssp585":
                ds = mf.open_files_CESM_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        elif model_name == "IPSL-CM6A-LR":
            if meta["scenario"] == "ssp585":
                ds = mf.open_files_IPSL_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        else:
            ds = mf.open_files(str(base))
        
        var = mf.read_var(ds, varname)
        
        # Calculate annual mean by year
        var_annual = mf.seasonal_mean_by_year(var, 1, 12)
        
        # Get coordinates
        lat = var_annual.lat.values
        lon = var_annual.lon.values
        
        # Create land/ocean masks using regionmask
        land_mask = None  # You can pass a pre-computed mask here if available
        
        if land_mask is None:
            try:
                import regionmask
                land_110 = regionmask.defined_regions.natural_earth_v5_0_0.land_110
                mask_np = land_110.mask(lon, lat)
                land_mask_da = xr.DataArray(~np.isnan(mask_np.values),
                                            dims=["lat", "lon"],
                                            coords={"lat": lat, "lon": lon})
            except ImportError:
                import warnings
                warnings.warn("regionmask not found — using all-ocean proxy.")
                land_mask_da = xr.DataArray(np.zeros((len(lat), len(lon)), dtype=bool),
                                            dims=["lat", "lon"],
                                            coords={"lat": lat, "lon": lon})
        else:
            land_mask_da = land_mask.interp(lat=lat, lon=lon, method="nearest") > 0.5
        
        ocean_mask_da = ~land_mask_da
        
        # Calculate global, land, and ocean means
        ts_global = var_annual.mean(dim=("lat", "lon"))
        ts_land = var_annual.where(land_mask_da).mean(dim=("lat", "lon"))
        ts_ocean = var_annual.where(ocean_mask_da).mean(dim=("lat", "lon"))
        
        # Store data
        data_by_experiment[exp]['global'].append({
            'model': model_name,
            'years': ts_global.year.values,
            'values': ts_global.values
        })
        data_by_experiment[exp]['land'].append({
            'model': model_name,
            'years': ts_land.year.values if hasattr(ts_land, 'year') else ts_global.year.values,
            'values': ts_land.values if hasattr(ts_land, 'values') else ts_land
        })
        data_by_experiment[exp]['ocean'].append({
            'model': model_name,
            'years': ts_ocean.year.values if hasattr(ts_ocean, 'year') else ts_global.year.values,
            'values': ts_ocean.values if hasattr(ts_ocean, 'values') else ts_ocean
        })

# Second pass: plot ensemble means and individual models
for ax, (exp, meta) in zip(axes, EXPERIMENTS.items()):
    if exp == "HIST":
        continue
    print(f"\nPlotting experiment: {exp}")
    
    # Plot individual models (thin lines)
    for model_data in data_by_experiment[exp]['global']:
        ax.plot(model_data['years'], model_data['values'], 
                color='black', alpha=0.3, linewidth=0.5)
    
    for model_data in data_by_experiment[exp]['land']:
        ax.plot(model_data['years'], model_data['values'], 
                color='red', alpha=0.3, linewidth=0.5)
    
    for model_data in data_by_experiment[exp]['ocean']:
        ax.plot(model_data['years'], model_data['values'], 
                color='blue', alpha=0.3, linewidth=0.5)
    
    # Calculate and plot ensemble means (thick lines)
    # Helper function to calculate ensemble mean across models with different time periods
    def calculate_ensemble_mean(model_data_list):
        if not model_data_list:
            return None, None
        
        # Find the common year range across all models
        min_year = max([m['years'].min() for m in model_data_list])
        max_year = min([m['years'].max() for m in model_data_list])
        common_years = np.arange(min_year, max_year + 1)
        
        # Interpolate each model to common years
        aligned_values = []
        for m in model_data_list:
            # Interpolate to common years
            interpolated = np.interp(common_years, m['years'], m['values'])
            aligned_values.append(interpolated)
        
        # Calculate ensemble mean
        ensemble_mean = np.mean(aligned_values, axis=0)
        return common_years, ensemble_mean
    
    # Global mean
    common_years, global_ensemble = calculate_ensemble_mean(data_by_experiment[exp]['global'])
    if global_ensemble is not None:
        ax.plot(common_years, global_ensemble, 
                color='black', linewidth=2, label='Global mean', zorder=10)
    
    # Land mean
    common_years, land_ensemble = calculate_ensemble_mean(data_by_experiment[exp]['land'])
    if land_ensemble is not None:
        ax.plot(common_years, land_ensemble, 
                color='red', linewidth=2, label='Land mean', zorder=10)
    
    # Ocean mean
    common_years, ocean_ensemble = calculate_ensemble_mean(data_by_experiment[exp]['ocean'])
    if ocean_ensemble is not None:
        ax.plot(common_years, ocean_ensemble, 
                color='blue', linewidth=2, label='Ocean mean', zorder=10)
    
    # Formatting
    ax.set_ylabel(f"{exp}\nTemperature (K)", fontsize=9)
    ax.set_ylim(270.0, 290.0)
    # ax.axhline(y=277, color='gray', linestyle='--', alpha=0.5)
    ax.axvspan(2021, 2050, color='lightpink', alpha=0.3)
    ax.axvspan(2071, 2100, color='lightblue', alpha=0.3)
    ax.grid(True, alpha=0.3)

# Final formatting
axes[-1].set_xlabel("Year", fontsize=10)
axes[0].legend(ncol=3, fontsize=9, loc="upper left")
plt.suptitle("Annual Mean Surface Temperature: Global, Land, and Ocean", y=0.93, fontsize=12)
plt.xlim(2015, 2100)
plt.tight_layout()

# Save
fig_directory = "/home/users/bidyut/UoE/20260112_Basic_Analysis"
figname = "Figure_Temperature_ByExperiment_GlobalLandOcean"
plt.savefig(f"{fig_directory}/{figname}.pdf", format="pdf", dpi=300)
print(f"\nFigure saved to: {fig_directory}/{figname}.pdf")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import warnings

# Define the two time periods
periods = {
    '2021-2050': (2021, 2050),
    '2071-2100': (2071, 2100)
}

# Storage for all model data by experiment and period
data_by_experiment_period = {
    period: {exp: {'global': [], 'land': [], 'ocean': []} 
             for exp in EXPERIMENTS.keys() if exp != "HIST"}
    for period in periods.keys()
}

# First pass: collect data from all models for each experiment
for model_name, model_meta in MODELS.items():
    print(f"Processing model: {model_name}")
    
    for exp, meta in EXPERIMENTS.items():
        if exp == "HIST":
            continue
            
        # --- special-case ensemble override ---
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ensemble = "r1i1p1f2"
            else:
                ensemble = "r1i1p1f1"
        else:
            ensemble = model_meta["ensemble"]
        
        base = (
            CEDA_BASE
            / meta["project"]
            / model_meta["institution"]
            / model_name
            / meta["scenario"]
            / ensemble
            / "Amon"
            / varname
            / model_meta["grid"]
            / "latest"
        )
        print(f"  {exp}: {str(base)}")
        
        # Open files with model-specific logic
        if model_name == "CESM2-WACCM":
            if meta["scenario"] == "G6sulfur":
                ds = mf.open_files_CESM_G6sulfur(base)
            elif meta["scenario"] == "ssp585":
                ds = mf.open_files_CESM_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        elif model_name == "IPSL-CM6A-LR":
            if meta["scenario"] == "ssp585":
                ds = mf.open_files_IPSL_ssp585(base)
            else:
                ds = mf.open_files(str(base))
        else:
            ds = mf.open_files(str(base))
        
        var = mf.read_var(ds, varname)
        tropics_latitude = 30
        var_tropics = var.sel(lat=slice(-tropics_latitude, tropics_latitude))
        
        # Calculate annual mean by year
        var_annual = mf.seasonal_mean_by_year(var_tropics, 1, 12)
        
        # Get coordinates
        lat = var_annual.lat.values
        lon = var_annual.lon.values
        
        # Create land/ocean masks using regionmask
        land_mask = None
        
        if land_mask is None:
            try:
                import regionmask
                land_110 = regionmask.defined_regions.natural_earth_v5_0_0.land_110
                mask_np = land_110.mask(lon, lat)
                land_mask_da = xr.DataArray(~np.isnan(mask_np.values),
                                            dims=["lat", "lon"],
                                            coords={"lat": lat, "lon": lon})
            except ImportError:
                warnings.warn("regionmask not found — using all-ocean proxy.")
                land_mask_da = xr.DataArray(np.zeros((len(lat), len(lon)), dtype=bool),
                                            dims=["lat", "lon"],
                                            coords={"lat": lat, "lon": lon})
        else:
            land_mask_da = land_mask.interp(lat=lat, lon=lon, method="nearest") > 0.5
        
        ocean_mask_da = ~land_mask_da
        
        # Calculate global, land, and ocean means
        ts_global = var_annual.mean(dim=("lat", "lon"))
        ts_land = var_annual.where(land_mask_da).mean(dim=("lat", "lon"))
        ts_ocean = var_annual.where(ocean_mask_da).mean(dim=("lat", "lon"))
        
        # Calculate period means for each time period
        for period_name, (start_year, end_year) in periods.items():
            # Select years in the period
            period_mask = (ts_global.year >= start_year) & (ts_global.year <= end_year)
            
            if period_mask.sum() > 0:  # Check if any years exist in this period
                global_mean = ts_global.where(period_mask, drop=True).mean().values
                land_mean = ts_land.where(period_mask, drop=True).mean().values
                ocean_mean = ts_ocean.where(period_mask, drop=True).mean().values
                
                # Store data
                data_by_experiment_period[period_name][exp]['global'].append({
                    'model': model_name,
                    'value': float(global_mean)
                })
                data_by_experiment_period[period_name][exp]['land'].append({
                    'model': model_name,
                    'value': float(land_mean)
                })
                data_by_experiment_period[period_name][exp]['ocean'].append({
                    'model': model_name,
                    'value': float(ocean_mean)
                })

# Create the bar plot
fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

# Experiments to plot (excluding HIST)
experiments_to_plot = [exp for exp in EXPERIMENTS.keys() if exp != "HIST"]

# Colors for the three categories
colors = {
    'global': 'black',
    'land': 'red',
    'ocean': 'blue'
}

# Bar width and spacing
bar_width = 0.25
group_spacing = 0.8  # Space between experiment groups

# Plot for each period
for ax_idx, (period_name, (start_year, end_year)) in enumerate(periods.items()):
    ax = axes[ax_idx]
    
    # X positions for each experiment
    x_positions = np.arange(len(experiments_to_plot)) * (3 * bar_width + group_spacing)
    
    for exp_idx, exp in enumerate(experiments_to_plot):
        x_base = x_positions[exp_idx]
        
        # Plot bars for global, land, ocean
        for cat_idx, (category, color) in enumerate(colors.items()):
            x_pos = x_base + cat_idx * bar_width
            
            # Get data for this experiment and category
            data = data_by_experiment_period[period_name][exp][category]
            
            if data:
                # Calculate ensemble mean
                values = [d['value'] for d in data]
                ensemble_mean = np.mean(values)
                
                # Plot bar
                ax.bar(x_pos, ensemble_mean, bar_width, 
                       color=color, alpha=0.6, edgecolor='black', linewidth=1)
                
                # Plot individual model points
                for value in values:
                    ax.plot(x_pos, value, 'o', color=color, 
                           markersize=6, markeredgecolor='black', 
                           markeredgewidth=0.5, alpha=0.8)
    
    # Formatting
    ax.set_xlabel('Experiment', fontsize=11, fontweight='bold')
    ax.set_title(f'{period_name}', fontsize=12, fontweight='bold')
    ax.set_xticks(x_positions + bar_width)
    ax.set_xticklabels(experiments_to_plot, rotation=0, ha='center')
    ax.grid(True, alpha=0.3, axis='y')
    ax.set_ylim(295.0, 305.0)
    ax.axhline(y=300, color='gray', linestyle='--', alpha=0.5, linewidth=1)

# Y-axis label only on left panel
axes[0].set_ylabel('Temperature (K)', fontsize=11, fontweight='bold')

# Create legend
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='black', alpha=0.6, edgecolor='black', label='Entire Tropics'),
    Patch(facecolor='red', alpha=0.6, edgecolor='black', label='Land Tropics'),
    Patch(facecolor='blue', alpha=0.6, edgecolor='black', label='Ocean Tropics')
]
axes[0].legend(handles=legend_elements, loc='upper left', fontsize=9)

# Overall title
fig.suptitle('Tropical Mean Surface Temperature by Experiment and Period', 
             fontsize=13, fontweight='bold', y=0.98)

plt.tight_layout()

# Save
fig_directory = "/home/users/bidyut/UoE/20260112_Basic_Analysis"
figname = "Figure_TropicalTemperature_Bars_TwoPeriods"
plt.savefig(f"{fig_directory}/{figname}.pdf", format="pdf", dpi=300, bbox_inches='tight')
print(f"\nFigure saved to: {fig_directory}/{figname}.pdf")
plt.show()
```

```python
import numpy as np
import xarray as xr

# ----------------------------------
# Robust hemispheric temperature diff
# ----------------------------------
def hemispheric_ts_diff(ts):
    """
    NH minus SH area-weighted mean surface temperature.
    ts: xarray.DataArray (lat, lon)
    """
    weights = np.cos(np.deg2rad(ts.lat))

    ts_nh = ts.sel(lat=slice(0, None))
    ts_sh = ts.sel(lat=slice(None, 0))

    w_nh = weights.sel(lat=slice(0, None))
    w_sh = weights.sel(lat=slice(None, 0))

    ts_nh_mean = ts_nh.weighted(w_nh).mean(("lat", "lon"))
    ts_sh_mean = ts_sh.weighted(w_sh).mean(("lat", "lon"))

    return ts_nh_mean - ts_sh_mean


# ----------------------------------
# Setup
# ----------------------------------
experiments = ["G6sulfur-SSP245", "G6solar-SSP245"]
models = list(ts_mean_all.keys())

TS_hemispheric_2071_2100 = xr.DataArray(
    np.full((len(experiments), len(models)), np.nan),
    coords={
        "experiment": experiments,
        "model": models
    },
    dims=("experiment", "model"),
    name="TS_NH_minus_SH"
)

# ----------------------------------
# Loop (FIXED indexing)
# ----------------------------------
for model in models:
    for exp in experiments:
        TS_hemispheric_2071_2100.loc[exp, model] = hemispheric_ts_diff(
            ts_mean_all[model][exp]
        )
```
