```python

%store -r bar_data_HIST_TRMM19852014_minus_TRMM 
# stored in 20260227_JJAS_NDJF_PR_BIAS-HIST-TRMM.ipynb

%store -r bar_data_G6sulfur20212100_minus_HIST19212000
%store -r bar_data_SSP58520212100_minus_HIST19212000
%store -r bar_data_SSP24520212100_minus_HIST19212000

%store -r bar_data_G6sulfur20212100_minus_SSP24520212100
%store -r bar_data_G6sulfur20212100_minus_SSP58520212100




import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import pickle
from pathlib import Path

# ── Load MSE data ────────────────────────────────────────────────────────────
SAVE_DIR = Path('/home/users/bidyut/UoE/20260218_Monsoons')
with open(SAVE_DIR / 'all_bar_data_Relative.pkl', 'rb') as f:
    all_bar_data = pickle.load(f)

# ── Load MSE_T data ────────────────────────────────────────────────────────────
with open(SAVE_DIR / 'all_bar_data_Relative_T.pkl', 'rb') as ft:
    all_bar_data_T = pickle.load(ft)

# ── Load MSE_Q data ────────────────────────────────────────────────────────────
with open(SAVE_DIR / 'all_bar_data_Relative_Q.pkl', 'rb') as fq:
    all_bar_data_Q = pickle.load(fq)

# ── Load MSE data ────────────────────────────────────────────────────────────
SAVE_DIR = Path('/home/users/bidyut/UoE/20260218_Monsoons')
with open(SAVE_DIR / 'evapotranspiration_bar_data_G6sulfur20212100_minus_SSP24520212100.pkl', 'rb') as fevap:
    all_bar_data_E = pickle.load(fevap)

# ── Load MSE_T data ────────────────────────────────────────────────────────────
with open(SAVE_DIR / 'regional_decomp_moisture_advection_change_G6sulfur20212100_minus_SSP24520212100.pkl', 'rb') as fqadv:
    all_bar_data_qadv = pickle.load(fqadv)
with open(SAVE_DIR / 'regional_decomp_thermodynamic_G6sulfur20212100_minus_SSP24520212100.pkl', 'rb') as fth:
    all_bar_data_th = pickle.load(fth)
with open(SAVE_DIR / 'regional_decomp_dynamic_G6sulfur20212100_minus_SSP24520212100.pkl', 'rb') as fdy:
    all_bar_data_dy = pickle.load(fdy)
with open(SAVE_DIR / 'regional_decomp_nonlinear_G6sulfur20212100_minus_SSP24520212100.pkl', 'rb') as fnl:
    all_bar_data_nl = pickle.load(fnl)

# ── Load MSE_Q data ────────────────────────────────────────────────────────────
with open(SAVE_DIR / 'MoistureRecycling_bar_data_G6sulfur20212100_minus_SSP24520212100.pkl', 'rb') as fqrecycle:
    all_bar_data_qrecycle = pickle.load(fqrecycle)



def scale_bar_data(data, factor):
    out = data.copy()
    out['means'] = np.array(out['means']) * factor
    out['per_model_all'] = [
        np.array(v) * factor for v in out['per_model_all']
    ]
    return out

    
#Load MSE change
bar_MSE_G6s_SSP245   = scale_bar_data(all_bar_data['MSEsfc_bar_data_G6sulfur20212100_minus_SSP24520212100'],0.01)

# Load T contribution to MSEsfc
bar_var3_G6s_SSP245   = scale_bar_data(all_bar_data_T['MSEsfc_bar_data_G6sulfur20212100_minus_SSP24520212100'],0.01)

# Load Q contribution to MSEsfc
bar_var4_G6s_SSP245   = scale_bar_data(all_bar_data_Q['MSEsfc_bar_data_G6sulfur20212100_minus_SSP24520212100'],0.01)

#Load Evap change
bar_E_G6s_SSP245   = scale_bar_data(all_bar_data_E['evapotranspiration_bar_data_G6sulfur20212100_minus_SSP24520212100'], 86400)

# Load Th_Dy contribution 
bar_qadv_G6s_SSP245 = scale_bar_data(all_bar_data_qadv, 8.64*10e8) ;#['moisture_advection_change']
bar_th_G6s_SSP245 = scale_bar_data(all_bar_data_th, 8.64*10e8) ;#['moisture_advection_change']
bar_dy_G6s_SSP245 = scale_bar_data(all_bar_data_dy, 8.64*10e8) ;#['moisture_advection_change']
bar_nl_G6s_SSP245 = scale_bar_data(all_bar_data_nl, 8.64*10e8) ;#['moisture_advection_change']

# Load Moisture reclin efficiency
bar_qrecycle_G6s_SSP245   = scale_bar_data(all_bar_data_qrecycle['MoistureRecycling_bar_data_G6sulfur20212100_minus_SSP24520212100'], 100)


# ── Shared styling ───────────────────────────────────────────────────────────
width     = 0.5
offsets   = np.array([-2.5, -1.5, -0.5, 0.5, 1.5, 2.5]) * width / 10
bar_color = ['gray', 'royalblue', 'forestgreen', 'gold', 'tomato', 'mediumpurple']
x = 0

panel_layout = [
    (0, 0, 'N. America'),
    (0, 1, 'Sahel'),
    (0, 2, 'India'),
    (1, 0, 'S. America'),
    (1, 1, 'S. Africa'),
    (1, 2, 'N. Australia'),
]

# ── Per-block dataset definitions ────────────────────────────────────────────
# Each block: list of (data_dict, offset, marker, color, label)
# None in data_dict position = blank bar slot

def make_datasets_rainfall():
    return [
        # (bar_data_HIST_TRMM19852014_minus_TRMM,          float(offsets[0]), 'o', bar_color[0], 'HIST−OBS'),
        (bar_data_G6sulfur20212100_minus_SSP24520212100,  float(offsets[4]), 'o', bar_color[4], 'G6sulfur−SSP245'),
    ]

def make_datasets(g6s_ssp245):
    return [
        (None,         float(offsets[0]), 'o', bar_color[0], 'HIST−OBS'),       # blank slot
        (g6s_ssp245,   float(offsets[4]), 'o', bar_color[4], 'G6sulfur−SSP245'),
    ]

blocks = [
    (make_datasets_rainfall(),                                                          '$\\Delta$P (mm day$^{-1}$)'),
    (make_datasets(bar_MSE_G6s_SSP245),  '$\\Delta$MSE$_{sfc}$ (J kg$^{-1}$)'),
    (make_datasets(bar_var3_G6s_SSP245), '$\\Delta$Temp. component'),
    (make_datasets(bar_var4_G6s_SSP245), '$\\Delta$Moist. component'),
    (make_datasets(bar_E_G6s_SSP245),    '$\\Delta$Evapotranspiration'),
    (make_datasets(bar_qadv_G6s_SSP245), '$\\Delta$Qadv'),
    (make_datasets(bar_th_G6s_SSP245), '$\\Delta$Qadv_Th'),
    (make_datasets(bar_dy_G6s_SSP245), '$\\Delta$Qadv_Dy'),
    (make_datasets(bar_nl_G6s_SSP245), '$\\Delta$Qadv_NL'),
    (make_datasets(bar_qrecycle_G6s_SSP245), '$\\Delta$Moisture Recycling Efficiency'),   
]

n_blocks  = len(blocks)   # 4
n_rows    = 2
n_cols    = 7

# ── Build figure with nested GridSpec ────────────────────────────────────────
# Outer GridSpec: 2 panel-rows × 4 panel-cols
# Inner GridSpec per cell: 4 sub-rows (one per variable block), no spacing

from matplotlib.gridspec import GridSpecFromSubplotSpec, GridSpec

fig, axes = plt.subplots(2, 3, figsize=(20, 10), sharey=False)
axes = axes.flatten()

all_labels     = list(bar_data_HIST_TRMM19852014_minus_TRMM['labels'])
monsoon_regions = ['N. America', 'Sahel', 'India', 'S. America', 'S. Africa', 'N. Australia']

for ax_idx, panel_label in enumerate(monsoon_regions):
    ax = axes[ax_idx]
    i_label = all_labels.index(panel_label)   # <-- correct lookup

    x_positions = np.arange(len(blocks))

    for b_idx, (datasets, ylabel) in enumerate(blocks):
        for data, offset, marker, color, leg_label in datasets:
            if data is None:
                continue

            val = data['means'][i_label]
            ax.bar(x_positions[b_idx] + offset, val,
                   width=width, color=color, edgecolor='k',
                   linewidth=0.7, alpha=0.85)

            pm_raw = np.array(data['per_model_all'][i_label])
            pm = pm_raw if pm_raw.ndim == 1 else pm_raw[:, -1]
            # ax.scatter(
            #     np.full(len(pm), x_positions[b_idx] + offset), pm,
            #     marker=marker, color='k', s=20,
            #     alpha=0.8, edgecolors='white', linewidths=0.4
            # )

    # Formatting — ONCE, inside loop
    ax.axhline(0, color='k', linestyle='--', linewidth=0.8)
    ax.set_xticks(x_positions)
    ax.set_xticklabels([b[1] for b in blocks], rotation=45, ha='right', fontsize=9)
    ax.set_title(panel_label, fontsize=14, fontweight='bold')
    ax.grid(axis='y', linestyle=':', linewidth=0.5, alpha=0.6)

plt.tight_layout()
plt.show()
```
